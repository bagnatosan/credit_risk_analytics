```python
df['person_emp_length'] = df['person_emp_length'].fillna(df['person_emp_length'].median()) 
df['loan_int_rate'] = df['loan_int_rate'].fillna(df['loan_int_rate'].median()) 

df.isnull().sum()
```

```python
df[(df['person_age'] > 100) | (df['person_emp_length'] > 60.0)]
```

### 🔍 Diagnóstico de Calidad del Dato y Limpieza

####  Tratamiento de Valores Nulos (Imputación):
* **`loan_int_rate`** ($3.116$ nulos, ~$9.5\%$): Se imputó con la **mediana** ($10.99\%$) por consistencia metodológica, coincidiendo estrechamente con la media ($11.01\%$) dada su distribución simétrica.
* **`person_emp_length`** ($895$ nulos, ~$2.7\%$): Se imputó con la **mediana** ($4.0$ años) para evitar el sesgo generado por registros con valores extremos.

####  Depuración de Errores de Registro (Outliers Críticos):
* **Detección:** Se identificaron $5$ registros con edades biológicamente imposibles (`person_age` de $123$ y $144$ años) y $2$ registros con antigüedad laboral inconsistente (`person_emp_length` de $123$ años en perfiles jóvenes).
* **Acción:** Se eliminaron un total de **$7$ registros** ($0.02\%$ del total de $32.581$), asegurando la integridad del dataset sin comprometer representatividad estadística.
